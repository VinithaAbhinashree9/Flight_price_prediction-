```python
# Simple Calculator in Python

# Take input from user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Choose operation
operation = input("Enter operation (+, -, *, /): ")

# Perform calculation
if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Cannot divide by zero"
else:
    result = "Invalid operation"

# Print result
print("Result:", result)
```

```python
import gradio as gr

def calculator(num1, num2, operation):
    if operation == "+":
        return num1 + num2
    elif operation == "-":
        return num1 - num2
    elif operation == "*":
        return num1 * num2
    elif operation == "/":
        if num2 != 0:
            return num1 / num2
        else:
            return "Cannot divide by zero"

interface = gr.Interface(
    fn=calculator,
    inputs=[
        gr.Number(label="Enter First Number"),
        gr.Number(label="Enter Second Number"),
        gr.Radio(["+", "-", "*", "/"], label="Select Operation")
    ],
    outputs="text",
    title="Simple Calculator with Gradio",
    description="A basic calculator built with Python and Gradio"
)

interface.launch()
```

```python
import gradio as gr

# Function to update display
def add_to_display(value, current):
    return current + value

# Function to clear display
def clear():
    return ""

# Function to calculate result
def calculate(expression):
    try:
        return str(eval(expression))
    except:
        return "Error"

with gr.Blocks() as calculator:
    gr.Markdown("## 🧮 Beautiful Gradio Calculator")

    display = gr.Textbox(label="Display", value="")

    with gr.Row():
        btn7 = gr.Button("7")
        btn8 = gr.Button("8")
        btn9 = gr.Button("9")
        btn_div = gr.Button("/")

    with gr.Row():
        btn4 = gr.Button("4")
        btn5 = gr.Button("5")
        btn6 = gr.Button("6")
        btn_mul = gr.Button("*")

    with gr.Row():
        btn1 = gr.Button("1")
        btn2 = gr.Button("2")
        btn3 = gr.Button("3")
        btn_sub = gr.Button("-")

    with gr.Row():
        btn0 = gr.Button("0")
        btn_dot = gr.Button(".")
        btn_eq = gr.Button("=")
        btn_add = gr.Button("+")

    with gr.Row():
        btn_clear = gr.Button("C")

    # Button actions
    buttons = [btn0,btn1,btn2,btn3,btn4,btn5,btn6,btn7,btn8,btn9,
               btn_add,btn_sub,btn_mul,btn_div,btn_dot]

    for b in buttons:
        b.click(add_to_display, inputs=[gr.Textbox(value=b.value, visible=False), display], outputs=display)

    btn_eq.click(calculate, inputs=display, outputs=display)
    btn_clear.click(clear, outputs=display)

calculator.launch()
```
