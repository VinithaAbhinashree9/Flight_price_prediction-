```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(data['price'], bins=20)
plt.title('Distribution of Flight Prices')  
```

```python
sns.scatterplot(x=data['departure_time'], y=data['duration'], hue=data['price'])
plt.title('Departure Time vs Duration')  
```

```python
sns.countplot(x=data['airline'])
plt.title('Count of Flights by Airline')    
```

```python
sns.barplot(x=data['airline'], y=data['price'])
plt.title('Average Price by Airline')
```

```python
sns.kdeplot(data['price'], shade=True)
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data['price'], bins=50, kde=True, color='steelblue')
plt.xlabel('Price (INR)')
plt.ylabel('Count')
plt.title('Flight Price Distribution')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data['price'])
plt.title('Box Plot - Price (Outlier Detection)')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='airline', data=data)
plt.title('Count of Flights per Airline')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
avg_price = data.groupby('airline')['price'].mean().sort_values(ascending=False)
avg_price.plot(kind='bar', color='coral', edgecolor='black')
plt.title('Average Price by Airline')
plt.ylabel('Average Price')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6, 6))
data['class'].value_counts().plot(kind='pie', autopct='%1.1f%%',
                                   colors=['skyblue', 'orange'])
plt.title('Flight Class Distribution')
plt.ylabel('')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='class', y='price', data=data)
plt.title('Price by Class')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data['price'], fill=True, color='green')
plt.title('KDE Plot - Price Distribution')
plt.show()
```
